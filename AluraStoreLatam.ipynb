### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingreso_tienda = tienda['Precio'].sum()
ingreso_tienda_2 = tienda2['Precio'].sum()
ingreso_tienda_3 = tienda3['Precio'].sum()
ingreso_tienda_4 = tienda4['Precio'].sum()
ingreso_tienda, ingreso_tienda_2, ingreso_tienda_3, ingreso_tienda_4
```

# 2. Ventas por categoría

```python
total_por_categoria = tienda.groupby('Categoría del Producto')['Precio'].sum()
total_por_categoria_2 = tienda2.groupby('Categoría del Producto')['Precio'].sum()
total_por_categoria_3 = tienda3.groupby('Categoría del Producto')['Precio'].sum()
total_por_categoria_4 = tienda4.groupby('Categoría del Producto')['Precio'].sum()
```

# 3. Calificación promedio de la tienda


```python
calificacion_promedio = {
    'Tienda 1': tienda['Calificación'].mean(),
    'Tienda 2': tienda2['Calificación'].mean(),
    'Tienda 3': tienda3['Calificación'].mean(),
    'Tienda 4': tienda4['Calificación'].mean()
}
tb_calificacion_promedio = pd.DataFrame.from_dict(calificacion_promedio, orient='index', columns=['Calificación promedio'])
tb_calificacion_promedio
```

# 4. Productos más y menos vendidos

```python
ventas_productos = tienda['Producto'].value_counts()
ventas_productos_2 = tienda2['Producto'].value_counts()
ventas_productos_3 = tienda3['Producto'].value_counts()
ventas_productos_4 = tienda4['Producto'].value_counts()
```

```python
cantidad_maxima = ventas_productos.max()
cantidad_minima = ventas_productos.min()
venta_mayor = ventas_productos[ventas_productos == cantidad_maxima].index.tolist()
venta_menor = ventas_productos[ventas_productos == cantidad_minima].index.tolist()
print(f'Tienda 1 \nProductos mas vendidos: \n {venta_mayor} \nProductos menos vendidos \n {venta_menor}')
```

```python
cantidad_maxima_2 = ventas_productos_2.max()
cantidad_minima_2 = ventas_productos_2.min()
venta_mayor_2 = ventas_productos_2[ventas_productos_2 == cantidad_maxima_2].index.tolist()
venta_menor_2 = ventas_productos_2[ventas_productos_2 == cantidad_minima_2].index.tolist()
print(f'Tienda 2 \nProductos mas vendidos: \n {venta_mayor_2} \nProductos menos vendidos \n {venta_menor_2}')
```

```python
cantidad_maxima_3 = ventas_productos_3.max()
cantidad_minima_3 = ventas_productos_3.min()
venta_mayor_3 = ventas_productos_3[ventas_productos_3 == cantidad_maxima_3].index.tolist()
venta_menor_3 = ventas_productos_3[ventas_productos_3 == cantidad_minima_3].index.tolist()
print(f'Tienda 3 \nProductos mas vendidos: \n {venta_mayor_3} \nProductos menos vendidos \n {venta_menor_3}')
```

```python
cantidad_maxima_4 = ventas_productos_4.max()
cantidad_minima_4 = ventas_productos_4.min()
venta_mayor_4 = ventas_productos_4[ventas_productos_4 == cantidad_maxima_4].index.tolist()
venta_menor_4 = ventas_productos_4[ventas_productos_4 == cantidad_minima_4].index.tolist()
print(f'Tienda 4 \nProductos mas vendidos: \n {venta_mayor_4} \nProductos menos vendidos \n {venta_menor_4}')
```
